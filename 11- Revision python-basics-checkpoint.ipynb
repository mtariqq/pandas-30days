```python
banks = [
    {"bank_name": "BAHL", "staff": 1500},
    {"bank_name": "MBL", "staff": 2000},
    {"bank_name": "Fbl", "staff": 800},
    {"bank_name": "ABl", "staff": 1800}
]

for bank in banks:
    print(bank["bank_name"], "-", bank["staff"])

total = 0
for bank in banks:
    total = total + bank["staff"]
print("total staff:", total)

max_staff = 0
biggest_bank = ""

for bank in banks:
    if bank["staff"] > max_staff:
        max_staff = bank["staff"]
        biggest_bank = bank["bank_name"]

print("Biggest bank:", biggest_bank, "with", max_staff, "staff")
```

```python
def print_trip_plan(plan):
    print("=== Your China Trip Plan ===")
    print(f"Main Base: {plan['destination']}")
    print(f"Total Days: {plan['days']}")

    print("\nCities to visit:") # this is our new list
    cities = plan.get("cities_to_visit", []) # get with empty list as default
    for i, city in enumerate(cities, start=1): #enumerate gives us 1,2,3...
        print(f"{i}, {city}")
    print(f"\nPurpose: {plan.get('purpose', 'Travel')}")
    print("Planned by:", plan.get("planner", "Guest"))

# New dict with a LIST inside
trip_plan = {
    "destination": "Hangzhou",
    "days": 3,
    "purpose": "interview perp + Travel",
    "cities_to_visit": ["West Lake", "Alibaba HQ", "Tea Village", "Qiantang River", "Zhejiang Unkiversity"], 
    # <-- List
    "planner": "Muhammad"
}

print_trip_plan(trip_plan)
    
```

```python
print("\n--- Trip Check ---")

if trip_plan["days"] >= 4:
    print("You have time for all cities")
    for i, city in enumerate(trip_plan["cities_to_visit"], start=1):
        print(f"{i}. {city}")

elif trip_plan["days"] >=2:
    print("short trip! Pick your top 2 cities:")
    for i, city in enumerate(trip_plan["cities_to_visit"][:2], start=1): # [:2] = first 2 only
        print(f"{i}. {city}")

else: 
    print("Too Short, for Hangzhou. Pick 1 city or postpone")
```

```python
trip_plan = {
    "destination": "Hangzhou",
    "days": 7,
    "cities_to_visit": [
    "west Lake - Scenic boat ride",
    "Alibaba HQ - Tech campus tour",
    "Tea Village - Dragonwell tea tasting",
    "Qiantang River - Tidal bore view",
    "Zhejiang University - Historic campus"
    ]
}

print(f"\n--- Hangzhou Trip: {trip_plan['days']} days ---")
print(f"Filtering for 'Lake' or 'River' spots only \n")

# Day 12 New Code: for + if together
for i, city in enumerate(trip_plan["cities_to_visit"], start=1):
    if "Lake" in city or "River" in city:
# Filter rule
        print(f"Day {i}: {city}")
```

```python
cities_to_visit = [
    "West Lake - Scenic boat ride",
    "Alibaba HQ - Tech campus tour",
    "Qiantang River - Tidal bore view",
    "Xixi Wetland - Nature walk",
    "Zhejiang University - Historic campus"
]

i = 0 # start counter
print("--- Filtering for 'Tech' or University spots ---")
      
while i < len(cities_to_visit): #loop while i is less than 5
    city = cities_to_visit[i] # get city at position i
    if "Tech" in city or "University" in city: #check rule
        print(f"Day {i+1}: {city}")
    i = i + 1 # move to next city

print("--- Filter complete ---")
```
